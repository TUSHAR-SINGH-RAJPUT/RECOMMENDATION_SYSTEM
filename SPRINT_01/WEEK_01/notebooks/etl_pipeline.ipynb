```python
import pandas as pd
import os
import logging
from datetime import datetime

# -----------------------------
# Setup logging
# -----------------------------
logging.basicConfig(
    filename="etl_log.log",
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

# -----------------------------
# File paths
# -----------------------------
BASE_DIR = os.path.dirname(__file__)
INPUT_FILE = os.path.abspath(os.path.join(BASE_DIR, "..", "user_behavior.csv"))
OUTPUT_FILE = os.path.join(BASE_DIR, "processed_user_behavior.csv")

# -----------------------------
# Extract
# -----------------------------
def extract():
    try:
        df = pd.read_csv(INPUT_FILE)
        logging.info("Data extracted successfully")
        return df
    except Exception as e:
        logging.error(f"Error in extract: {e}")
        raise

# -----------------------------
# Transform
# -----------------------------
def transform(df):
    try:
        df.columns = df.columns.str.strip()

        # Drop missing values
        df = df.dropna()

        # Convert types
        df["user_id"] = df["user_id"].astype(int)
        df["product_id"] = df["product_id"].astype(int)

        # Map interaction weights
        interaction_weights = {
            "view": 1,
            "cart": 2,
            "purchase": 3
        }

        df["interaction_score"] = df["interaction_type"].map(interaction_weights)

        # Remove invalid rows
        df = df.dropna(subset=["interaction_score"])

        logging.info("Data transformed successfully")
        return df

    except Exception as e:
        logging.error(f"Error in transform: {e}")
        raise

# -----------------------------
# Load
# -----------------------------
def load(df):
    try:
        df.to_csv(OUTPUT_FILE, index=False)
        logging.info("Data loaded successfully")
    except Exception as e:
        logging.error(f"Error in load: {e}")
        raise

# -----------------------------
# Pipeline Runner
# -----------------------------
def run_pipeline():
    logging.info("ETL Pipeline Started")

    df = extract()
    df = transform(df)
    load(df)

    logging.info("ETL Pipeline Completed Successfully")
    print("✅ ETL Pipeline Completed")

# -----------------------------
# Run
# -----------------------------
if __name__ == "__main__":
    run_pipeline()
```
